# AŞAMA 4: Veri Temizliğinin Tamamlanması

```python
import pandas as pd

df = pd.read_csv("credit_risk_dataset.csv")

onceki_boyut = df.shape
print("Temizlik öncesi:", onceki_boyut)
```

### 1. Eksik Veri Kontrolü

```python
print(df.isnull().sum())
```

```python
df["person_emp_length"] = df["person_emp_length"].fillna(df["person_emp_length"].median())
df["loan_int_rate"] = df["loan_int_rate"].fillna(df["loan_int_rate"].median())

print(df.isnull().sum().sum())
```

### 2. Tekrar Kontrolü

```python
print("Tekrar eden satır sayısı:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Tekrar eden satır sayısı:", df.duplicated().sum())
```

### 3. Sınır / Aykırı Değer Kontrolü

```python
df.describe()
```

`person_age` sütununda en büyük değer 144, `person_emp_length` sütununda ise 123 yıl. Bunlar gerçek olamayacak değerler olduğu için yaşı 100'den, çalışma süresi 60 yıldan büyük kayıtlar filtrelendi.

```python
print(df[(df["person_age"] > 100) | (df["person_emp_length"] > 60)])
```

```python
df = df[(df["person_age"] <= 100) & (df["person_emp_length"] <= 60)]

print("En büyük yaş:", df["person_age"].max())
print("En büyük çalışma süresi:", df["person_emp_length"].max())
```

### 4. Boyut Doğrulaması

```python
print("Temizlik öncesi:", onceki_boyut)
print("Temizlik sonrası:", df.shape)
print("Silinen satır sayısı:", onceki_boyut[0] - df.shape[0])
```

```python
df.to_csv("credit_risk_dataset_temiz.csv", index=False)
```

### 5. Markdown Özeti

**Bu veri setinde hangi temizlik adımlarını yaptım ve neden?**

Çalışma süresi ve faiz oranı sütunlarındaki eksik değerleri, aykırı değerlerden etkilenmemesi için medyan ile doldurdum; satır silseydim verinin yaklaşık %12'sini kaybedecektim. Analiz sonuçlarını bozmaması için 165 tekrarlanan kaydı sildim. Ayrıca 144 yaş ve 123 yıl çalışma süresi gibi gerçek olamayacak 7 kaydı filtreledim ve veri seti 32.581 satırdan 32.409 satıra indi.
