# AŞAMA 3: Veri Ön İşleme

```python
import pandas as pd

df = pd.read_csv("credit_risk_dataset.csv")
```

### 1. Durum Tespiti

```python
print(df.isnull().sum())
```

```python
print("Tekrar eden satır sayısı:", df.duplicated().sum())
```

Veri setinde `person_emp_length` sütununda 895, `loan_int_rate` sütununda 3116 eksik değer ve 165 tekrarlanan kayıt bulunmaktadır.

### 2. Temizleme / Doldurma

```python
df = df.drop_duplicates()

df["person_emp_length"] = df["person_emp_length"].fillna(df["person_emp_length"].median())
df["loan_int_rate"] = df["loan_int_rate"].fillna(df["loan_int_rate"].median())
```

```python
print(df.isnull().sum())
print("Tekrar eden satır sayısı:", df.duplicated().sum())
```

Tekrarlanan kayıtlar `drop_duplicates()` ile silindi, eksik değerler `fillna()` ile sütunların medyan değeriyle dolduruldu. Temizlik sonrası veri setinde eksik ve tekrarlanan kayıt kalmamıştır.
