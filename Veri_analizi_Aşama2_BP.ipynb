## NumPy ile İnceleme

**Amaç:** Kredi tutarı, faiz oranı ve yaş sütunlarını NumPy dizilerine çevirip temel istatistiklerini hesaplamak.

```python
import numpy as np

tutarlar = df["loan_amnt"].to_numpy()
faizli_tutar = tutarlar * 1.10

print(tutarlar[:5])
print(faizli_tutar[:5])
```

```python
print("Toplam:", round(np.sum(tutarlar),2))
print("Ortalama:", round(np.mean(tutarlar),2))
print("Medyan:", round(np.median(tutarlar),2))
print("En küçük:", round(np.min(tutarlar),2))
print("En büyük:", round(np.max(tutarlar),2))
print("Standart sapma:", round(np.std(tutarlar),2))
```

### Faiz Oranındaki Eksik Değerler

* **np.mean:** Dizide eksik değer (NaN) varsa sonuç da NaN çıkar.
* **np.nanmean:** Eksik değerleri yok sayarak ortalamayı hesaplar.
* **np.isnan().sum():** Kaç tane eksik değer olduğunu sayar.

```python
faizler = df["loan_int_rate"].to_numpy()

print("Eksik değer sayısı:", np.isnan(faizler).sum())
print("np.mean ile:", np.mean(faizler))
print("np.nanmean ile:", round(np.nanmean(faizler),2))
```

```python
yaslar = df["person_age"].to_numpy()

print("Ortalama yaş:", round(np.mean(yaslar),2))
print("En büyük yaş:", np.max(yaslar))
print("100 yaşından büyük olanlar:", yaslar[yaslar > 100])
```

**Çıkarımım:** Yaş sütununda 123 ve 144 gibi gerçek olamayacak değerler var. Bunlar büyük ihtimalle veri girişi hatası.

```python
tablo = df[["loan_amnt", "person_income"]].to_numpy()

print(tablo[:3])
print(tablo.mean(axis=0))
print(tablo[:3].sum(axis=1))
```

### Gelir Grubu Oluşturma (np.where)

**Amaç:** Medyan gelirin üstünde ve altında kalan kişileri iki gruba ayırmak.

```python
gelirler = df["person_income"].to_numpy()

gelir_grubu = np.where(gelirler >= np.median(gelirler), "Yüksek", "Düşük")

print(np.unique(gelir_grubu, return_counts=True))
```

## Kredi Durumunun Dağılımı

**Amacımız:** Kredilerin ne kadarının geri ödendiğini (0), ne kadarının temerrüde düştüğünü (1) görmek.

**Çıkarımım:** Kredilerin yaklaşık %78'i geri ödenmiş, %22'si temerrüde düşmüş.

```python
print(df["loan_status"].value_counts(normalize=True) * 100)
```

## Kredi Amaçlarının Dağılımı

**Amacımız:** Kredilerin en çok hangi amaçla alındığını görmek.

```python
print(df["loan_intent"].value_counts(normalize=True) * 100)
```

## Kredi Notuna Göre Temerrüt Oranı

**Amaç:** Kredi notu (A'dan G'ye) kötüleştikçe temerrüt oranının ve faizin nasıl değiştiğini görmek.

**Yöntem:** `groupby` ile kredi notuna göre gruplayıp `loan_status` ve `loan_int_rate` ortalamaları alındı.

**Sonuç:** A notunda temerrüt oranı yaklaşık %10 iken, not kötüleştikçe oran ve faiz düzenli olarak artıyor. Kredi notu, riski iyi yansıtıyor.

```python
print(df.groupby("loan_grade")[["loan_status", "loan_int_rate"]].mean())
```

## Kredi Amacına Göre Temerrüt Oranı

**Çıkarımım:** En riskli amaç borç kapatma (DEBTCONSOLIDATION), en az riskli amaç ise girişim (VENTURE) kredileri.

```python
print(df.groupby("loan_intent")["loan_status"].mean().sort_values(ascending=False) * 100)
```

## Konut Durumuna Göre İnceleme

**Çıkarımım:** Kirada oturanların (RENT) temerrüt oranı, ev sahibi olanlara (OWN) göre çok daha yüksek.

```python
print(df.groupby("person_home_ownership")[["person_income", "loan_status"]].mean())
```

## Geçmiş Temerrüt Kaydı ve Mevcut Kredi Durumu

**Amaç:** Daha önce temerrüde düşmüş kişilerin tekrar temerrüde düşüp düşmediğini görmek.

**Sonuç:** Geçmişte temerrüt kaydı olanların (Y) yaklaşık %38'i tekrar temerrüde düşmüş. Kaydı olmayanlarda (N) bu oran yaklaşık %18.

```python
print(pd.crosstab(df["cb_person_default_on_file"], df["loan_status"], normalize="index") * 100)
```

## Veri Temizleme

**Amaç:** Veri setini analize hazır hale getirmek.

**Yöntem:**
* Tekrar eden satırlar silindi.
* Yaşı 100'den, çalışma süresi 60 yıldan büyük olan hatalı kayıtlar çıkarıldı.
* Eksik değerler sütunun medyanı ile dolduruldu.

```python
df_temiz = df.drop_duplicates()

df_temiz = df_temiz[df_temiz["person_age"] <= 100]
df_temiz = df_temiz[(df_temiz["person_emp_length"] <= 60) | (df_temiz["person_emp_length"].isnull())].copy()

df_temiz["person_emp_length"] = df_temiz["person_emp_length"].fillna(df_temiz["person_emp_length"].median())
df_temiz["loan_int_rate"] = df_temiz["loan_int_rate"].fillna(df_temiz["loan_int_rate"].median())

print(df.shape)
print(df_temiz.shape)
print(df_temiz.isnull().sum().sum())
```

```python
df_temiz.to_csv("credit_risk_dataset_temiz.csv", index=False)
```

# AŞAMA 2 – Düzenlenmiş Veri Seti Üzerinde Analiz

Bu aşamada temizlenmiş veri seti (`credit_risk_dataset_temiz.csv`) kullanıldı.

```python
df_temiz = pd.read_csv("credit_risk_dataset_temiz.csv")

print(df_temiz.shape)
```

## 1. Frekans – value_counts()

**Amaç:** Kredilerin hangi amaçlarla alındığını ve her amaçtan kaç kredi olduğunu görmek.

```python
print(df_temiz["loan_intent"].value_counts())
```

**Çıkarımım:** En çok kredi eğitim (EDUCATION) amacıyla alınmış, onu sağlık (MEDICAL) takip ediyor. En az kredi ise ev tadilatı (HOMEIMPROVEMENT) için alınmış.

## 2. Sıralama – sort_values()

**Amaç:** Yıllık geliri en yüksek olan ilk 5 kişiyi listelemek.

```python
print(df_temiz.sort_values("person_income", ascending=False).head(5))
```

**Çıkarımım:** Geliri en yüksek 5 kişinin hepsi kredisini geri ödemiş (loan_status = 0). Bu kişilerin aldığı kredi tutarları gelirlerine göre çok küçük, bu yüzden geri ödemede zorlanmamışlar.

## 3. Gruplama – groupby()

**Amaç:** Kredi notuna (loan_grade) göre ortalama faiz oranını (loan_int_rate) görmek.

```python
print(df_temiz.groupby("loan_grade")["loan_int_rate"].mean().round(2))
```

## 4. Yorum

En dikkat çekici bulgu, kredi notu A'dan G'ye doğru kötüleştikçe ortalama faiz oranının her adımda düzenli olarak artmasıdır: A notunda faiz yaklaşık %7,67 iken G notunda %19,53'e çıkıyor. Yani bankalar riskli gördükleri müşterilere neredeyse 2,5 kat daha yüksek faiz uyguluyor.
